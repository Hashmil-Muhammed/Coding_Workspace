# 1. Magic Square

## What is a Magic Square?

A **Magic Square** is a square matrix of order **N × N** that contains all integers from **1 to N²** exactly once.

The arrangement is done in such a way that:

- Every row has the same sum.
- Every column has the same sum.
- Both diagonals have the same sum.

This common sum is called the **Magic Constant**.

---

## Example

For **N = 3**

```
2 7 6
9 5 1
4 3 8
```

### Row Sum

```
2 + 7 + 6 = 15
9 + 5 + 1 = 15
4 + 3 + 8 = 15
```

### Column Sum

```
2 + 9 + 4 = 15
7 + 5 + 3 = 15
6 + 1 + 8 = 15
```

### Diagonal Sum

```
2 + 5 + 8 = 15

6 + 5 + 4 = 15
```

Hence, it is called a **Magic Square**.

---

## Magic Constant Formula

The sum of every row, column, and diagonal is called the **Magic Constant**.

Formula


    n(n²+1) / 2                                             


Example

For **N = 3**

3*(9+1) / 2
= 30/2 = 15


## Approach Used

For odd values of **N**, the **Siamese Algorithm** is used.

Steps:

1. Start from the first row and middle column.
2. Place number **1**.
3. Move one row up and one column right.
4. If you move outside the matrix, wrap around.
5. If the next position is already occupied, move one row down instead.
6. Repeat until all numbers are placed.

---

## Algorithm

**Step 1**

Create an empty **N × N** matrix.

**Step 2**

Initialize

- Row = 0
- Column = N // 2

**Step 3**

Place numbers from **1** to **N²**

**Step 4**

For every new number

- Move Up
- Move Right

**Step 5**

If row becomes **-1**

Move to the last row.

**Step 6**

If column becomes **N**

Move to the first column.

**Step 7**

If the position is already filled

Move one row down from the previous position.

Repeat until all numbers are placed.

````python
n= int(input())
a = [[0] * n for _ in range(n)]

i, j = 0, n // 2
for x in range(1, n * n + 1):
    a[i][j] = x
    ni, nj = (i - 1) % n, (j + 1) % n
    if a[ni][nj]:
        i = (i + 1) % n
    else: 
        i, j = ni, nj

for r in a:
    print(*r)
````
