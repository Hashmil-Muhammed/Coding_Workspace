# **1. Reverse a String**
---

````python
s = "hello"
reverse = s[::-1]
print(reverse)
````

````python
def reverse(s):
    return s[::-1]
print(reverse("hello"))
````

 **How does [::-1] work?**

Python slicing syntax:
> string[start:stop:step]

when we write 
>s[::-1]

it means 
- start -> beginning
- stop -> end
- step -> -1


### **Without slicing**

````python
s= "hello"
reverse = ""
for i in range(len(s) -1, -1, -1):
    reverse += s[i]
print(reverse)
````

+= means add to the existing value.

**Interview explanation**

    "I start from the last character of the string and iterate backwards until the first character, appending each character to a new string."

**Complexity**
>Time:  O(n)

>Space: O(n)

### **Reverse in one-liner**

````python
print("hello"[::-1])
````

# **2. Swap 2 Numbers Without Using a Third Variable**

---

Suppose:
>a = 10

>b = 20

We want:
>a = 20

>b = 10


````python
a = 10
b= 20
a,b = b,a

print("a=",a)
print("b=",b)
````

This is called tuple unpacking.

Python internally allows us to assign:

    a, b = b, a

without explicitly creating a third variable.

**Arithmetic method**

````python
a =10
b= 20

a= a+b
b= a-b
a=a-b

print(a)
print(b)
````

Let's understand:

Initially:

>a = 10

>b = 20

>Step 1

>a = a + b


Now:

>a = 30

>b = 20

>Step 2

>b = a - b

>b = 30 - 20

>b = 10


Now:

>a = 30

>b = 10

>Step 3

>a = a - b

>a = 30 - 10

>a = 20


Final:

>a = 20

>b = 10


**Interview answer**

    "In Python, I can swap two variables using tuple unpacking with a, b = b, a, so I don't need an explicit third variable."

**Important**

If they specifically ask:

    "Do it without a third variable and without Python's built-in swapping?"

Use the arithmetic approach.

## **Swap 2 Numbers using a Third Variable**

````python
a = 10
b = 20

c = a
a=b
b=c

print(a)
print(b)
````

Simple logic:

1. c = a → save a
1. a = b → put b into a
1. b = c → put old a into b

So 10 and 20 become 20 and 10.

👉 c is the third variable used to temporarily store a.

**What is an Array?**

```An array is a linear data structure that stores multiple elements of the same type in a continuous memory location, where each element can be accessed using an index.```

``Array = Multiple + Linear + Index + Mutable + Fast Access``

Instead of:

    a = 10
    b = 20
    c = 30

We can use an array/list:

    numbers = [10, 20, 30]

Here, numbers stores 3 values.

1. Multiple elements

2. Linear Data Structure
An array is a linear data structure because elements are arranged one after another.

    Examples of linear data structures:

    - Array
    - Linked List
    - Stack
    - Queue

    Examples of non-linear data structures:

    - Tree
    - Graph

3. Indexing
Each element has an index.

    Think of it like a box 📦
    ```
    numbers = [10, 20, 30, 40]
    ```

    You can access values using the index:

        print(numbers[0])  # 10

4. Is an Array Mutable?

    Yes, an array is generally mutable.

    Mutable means we can change its values after creating it.

    For example, in Python:

        numbers = [10, 20, 30]
        numbers[1] = 50
        print(numbers)

    Output:
        [10, 50, 30]

    We changed 20 → 50.

    So:

    Mutable = Can be changed.

        [10, 20, 30, 40]

    Array = A collection of multiple values stored under one variable, accessed using an index.

Here, "array" may mean a string/character array.

For example:

```Input: "hello world"

Input: "hello world"

Vowels: e, o, o

Count = 3
```

````python
arr = ['h', 'e', 'l', 'l', 'o']
vowels = "aeiou"
count = 0

for i in arr:
    if i.lower() in vowels:
        count += 1
print(count)
````

**working:**

    arr = ['h', 'e', 'l', 'l', 'o']
    vowels = "aeiou"
    count = 0

arr → contains the letters.
vowels → contains a, e, i, o, u.
count = 0 → starts counting from 0.

Loop:

    for i in arr:

Takes each letter one by one:

h → e → l → l → o

Then:

    if i.lower() in vowels:

Checks whether the letter is a vowel.

- h → ❌
- e → ✅ → count = 1
- l → ❌
- l → ❌
- o → ✅ → count = 2

Finally:

    print(count)


👉 There are 2 vowels: e and o.


**How to explain it**

    "I iterate through every character and check whether it belongs to the set of vowels aeiou. If it is a vowel, I increment the counter."

**Complexity**

If there are n characters:

    Time: O(n)
    Space: O(1)

# **4. Frequency of Each Character in an Array**
---

commonly use this to test dictionaries/hash maps.

Example:

Input:

    banana

Expected:

    b → 1
    a → 3
    n → 2

````python
arr = ['b', 'a', 'n', 'a', 'n', 'a']

count = {} #Empty dictionary

for i in arr:
    if i in count:
        count[i] += 1
    else:
        count[i] = 1
print(count)
````

**Working**

    count = {}

Creates an empty dictionary to store character and count.

    for i in arr:

Takes each character one by one: b → a → n → a → n → a

    if i in count:
        count[i] += 1

If the character already exists, increase its count by 1.

    else:
        count[i] = 1

If the character is new, start its count at 1.

Working:
- b → {'b': 1}
- a → {'b': 1, 'a': 1}
- n → {'b': 1, 'a': 1, 'n': 1}
- a → {'b': 1, 'a': 2, 'n': 1}
- n → {'b': 1, 'a': 2, 'n': 2}
- a → {'b': 1, 'a': 3, 'n': 2}


**Interview explanation**

    "I use a dictionary as a hash map where each character is the key and its occurrence count is the value. For every character, I increment its existing count or initialize it to one."

**Complexity**

    Time: O(n)
    Space: O(k)

where k is the number of unique characters.

# **5. Prime Number Checking**

````python
num = 7
if num < 2:
     print("not prime")
else:
    for i in range(2, num):
        if num % i == 0:
            print("not prime")
            break
    else: 
        print("prime")
````

**Simple Explanation**

    num = 7

We want to check whether 7 is prime.

    if num < 2:
        print("Not Prime")

Numbers less than 2 are not prime.

    for i in range(2, num):

Checks numbers from 2 to 6.

    if num % i == 0:

% gives the remainder.

If remainder is 0, the number is exactly divisible → Not Prime.

    break

Stops the loop when a divisor is found.

    else:
        print("Prime")

If no divisor is found, print Prime.

For 7:
7 % 2 ≠ 0
7 % 3 ≠ 0
7 % 4 ≠ 0
7 % 5 ≠ 0
7 % 6 ≠ 0

7 % 7 = 1

No exact division → Prime ✅

**Make Prime Checking a Function**

````python
def check_prime(num):
    if num < 2:
        return False
    for i in range(2, num):
        if num % i == 0:
            return False 
    return True
print(check_prime(7))
````

**Simple working:**
- def check_prime(num): → creates a function.
- num < 2 → not prime → False
- Loop checks whether num is divisible by any number.
- If divisible → False
- If not divisible by anything → True
