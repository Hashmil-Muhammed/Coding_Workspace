# **3. Check prime number**

**1. Using simple code**

```python
num = 7
if num < 2:
     print("not prime")
else:
    for i in range(2, num):
        if num % i == 0:
            print("not prime")
            break
    else: 
        print("prime")
```

**2. Using a loop**

```python
n = 7
prime = True

for i in range(2, n):
    if n % i == 0:
        prime = False
        break

if n > 1 and prime:
    print("Prime")
else:
    print("Not Prime")
```

**3. Using a function**

```python
def prime(n):
    if n <= 1:
        return False

    for i in range(2, n):
        if n % i == 0:
            return False

    return True

print("Prime" if prime(7) else "Not Prime")
```

# **13. Armstrong Number**

**1. Simple code**

```python
n = 153
total = 0

for digit in str(n):
    total += int(digit) ** 3

print("Armstrong" if total == n else "Not Armstrong")
```

**2. Using a loop**

```python
n = 153
temp = n
total = 0

while n > 0:
    digit = n % 10
    total += digit ** 3
    n //= 10

print("Armstrong" if total == temp else "Not Armstrong")
```

**3. Using a function**

```python
def armstrong(n):
    total = 0

    for digit in str(n):
        total += int(digit) ** 3

    return total == n

print("Armstrong" if armstrong(153) else "Not Armstrong")
```

# **26. Find Second Most Repeated Character**

**1. Simple code**

```python
s = "aabbbcc"
print(sorted(set(s), key=s.count, reverse=True)[1])
```

**2. Using a loop**

```python
s = "aabbbcc"
freq = {}

for ch in s:
    freq[ch] = freq.get(ch, 0) + 1

result = sorted(freq, key=freq.get, reverse=True)
print(result[1])
```

**3. Using a function**

```python
def second_most(s):
    freq = {}

    for ch in s:
        freq[ch] = freq.get(ch, 0) + 1

    result = sorted(freq, key=freq.get, reverse=True)
    return result[1]

print(second_most("aabbbcc"))
```

# **29. GCD of Two Numbers**

**1. Simple code**

```python
import math

print(math.gcd(12, 18))
```

**2. Using a loop**

```python
a = 12
b = 18

while b != 0:
    a, b = b, a % b

print(a)
```

**3. Using a function**

```python
def gcd(a, b):
    while b != 0:
        a, b = b, a % b
    return a

print(gcd(12, 18))
```

# **30. Find Common Characters in Two Strings**

**1. Simple code**

```python
a = "hello"
b = "world"

print(set(a) & set(b))
```

**2. Using a loop**

```python
a = "hello"
b = "world"

for ch in a:
    if ch in b:
        print(ch)
```

**3. Using a function**

```python
def common_chars(a, b):
    result = []

    for ch in a:
        if ch in b and ch not in result:
            result.append(ch)

    return result

print(common_chars("hello", "world"))
```
